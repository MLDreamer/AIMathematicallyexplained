```python
import torch
import torch.nn as nn

# The pre-trained weights become our prior mean
theta_0 = torch.tensor([2.0], requires_grad=False)

# Prior precision (1 / sigma^2). Higher = stronger prior.
prior_precision = 1.0

# Initialize fine-tuning at the pre-trained weights
theta = torch.tensor([2.0], requires_grad=True)

# Fine-tuning data (likelihood will peak at 5.0)
x = torch.tensor([5.0])

optimizer = torch.optim.SGD([theta], lr=0.01)

for step in range(1000):
    optimizer.zero_grad()

    # Negative log likelihood (the cross entropy loss)
    nll = 0.5 * (theta - x).pow(2).sum()

    # Negative log prior (the L2 regularizer toward theta_0)
    neg_log_prior = 0.5 * prior_precision * (theta - theta_0).pow(2).sum()

    # MAP loss = negative log posterior
    loss = nll + neg_log_prior
    loss.backward()
    optimizer.step()

print(f"MAP estimate: {theta.item():.4f}")
```
